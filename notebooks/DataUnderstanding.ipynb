#### 1. Which column must be excluded because it is an identifier?

CustomerID

It is a unique identifier for each customer and does not contain meaningful information that should be learned by the model.

#### 2. Why should Future12MRevenueEGP not be used to predict current churn?

Because Future12MRevenueEGP contains future information.

We are trying to predict whether a customer will churn based on information available at the current time. Using future revenue would introduce data leakage because the model would have access to information that would not be available when making the churn prediction.

#### 3. Is accuracy sufficient for churn prediction?

No.

Accuracy alone can be misleading, especially when the churn classes are imbalanced.

We should also evaluate:

Precision
Recall
F1-score
Confusion Matrix

For churn, Recall can be particularly important because we want to identify as many actual churners as possible.

#### 4. Which algorithms are sensitive to feature scaling?

Algorithms that depend on distances or feature magnitude are sensitive to scaling, such as:

K-Nearest Neighbors (KNN)
SVM
Logistic Regression
Neural Networks
K-Means

Tree-based models such as:

Decision Tree
Random Forest
Gradient Boosting

generally do not require feature scaling.

### Part B — Data Preprocessing

```python
X = df.drop(columns=[
    "CustomerID",
    "Future12MRevenueEGP",
    "Churn"
])

y = df["Churn"]
```

```python
num_cols = [
    "Age",
    "TenureMonths",
    "MonthlyUsageGB",
    "MonthlyChargeEGP",
    "NumServices",
    "SupportCalls6M",
    "LatePayments12M",
    "SatisfactionScore"
]
```

```python
cat_cols = [
    "Region",
    "ContractType",
    "InternetType",
    "PaperlessBilling",
    "AutoPay"
]
```

##### Split the Data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y,
    shuffle=True
)
```

##### Median imputation + standardization


```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

##### Most-frequent imputation + One-Hot Encoding + handle_unknown="ignore"

```python
from sklearn.preprocessing import OneHotEncoder

cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

##### Combine the Pipelines

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python
print("Training shape:", X_train_processed.shape)
print("Testing shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(random_state=42))
])
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, pos_label="Yes"))
print("Recall:", recall_score(y_test, y_pred, pos_label="Yes"))
print("F1 Score:", f1_score(y_test, y_pred, pos_label="Yes"))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

##### 
706 → correctly predicted non-churn //
31 → non-churn predicted as churn  //
136 → churn predicted as non-churn  //
27 → correctly predicted churn  //

The important issue is the low recall (16.6%). The model is missing many actual churners, even though its accuracy is about 81.4%. This is exactly why accuracy alone isn't enough for this problem.

### Part C — Churn Classification

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    RocCurveDisplay
)

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "KNN": KNeighborsClassifier(),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        random_state=42
    ),

    "SVM": SVC(
        probability=True,
        random_state=42
    ),

    "Naive Bayes": GaussianNB(),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    )
}
```

```python
results = []
trained_models = {}

for name, classifier in models.items():

    model = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", classifier)
    ])

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test,
            y_pred,
            pos_label="Yes"
        ),
        "Recall": recall_score(
            y_test,
            y_pred,
            pos_label="Yes"
        ),
        "F1": f1_score(
            y_test,
            y_pred,
            pos_label="Yes"
        ),
        "ROC-AUC": roc_auc_score(
            y_test,
            y_prob
        )
    })

    trained_models[name] = model
```

```python
results_df = pd.DataFrame(results)

results_df.round(3)
```

#####
// Naive Bayes has the best F1 (0.447), by far the best recall (0.577) and the top ROC-AUC (0.748). Its accuracy is lower (0.741) because it flags many customers as churners, but for a retention campaign that is usually an acceptable trade.      
// Logistic Regression has essentially the same AUC (0.745), so its ranking ability is equivalent. At the default 0.5 threshold it is too conservative (recall 0.166), so it would benefit most from threshold tuning.          
// Random Forest and Gradient Boosting tie on AUC (0.726) with low recall. Their default predictions are also too conservative.               
// Decision Tree has the second-best recall (0.362) but the second-worst AUC (0.579), so its probability ranking is weak and it is likely overfitting.                
// KNN is the weakest overall (AUC 0.611).
